Задача #1
```
Брошены три монеты. Найти вероятности событий:
а) A={первая монета выпала гербом};
б) B={выпало ровно два герба};
в) C={выпало не больше двух гербов}.
```

````python
import numpy as np

np.random.seed(42)
n = 100000  # число симуляций

# 0 — решка, 1 — герб
coins = np.random.randint(0, 2, (n, 3))

# а) Первая монета — герб
A = coins[:, 0] == 1
P_A = np.mean(A)

# б) Ровно два герба
two_heads = np.sum(coins, axis=1) == 2
P_B = np.mean(two_heads)

# в) Не больше двух гербов
no_more_than_two = np.sum(coins, axis=1) <= 2
P_C = np.mean(no_more_than_two)

print(f"а) Вероятность A (первая — герб): {P_A:.3f}")
print(f"б) Вероятность B (ровно два герба): {P_B:.3f}")
print(f"в) Вероятность C (не больше двух гербов): {P_C:.3f}")
````

Аналитические решения:
```
а)0.5
б)0.375
в)0.875
```



Задача #2
```
В лифт 9-этажного дома на первом этаже вошли 5 человек. Известно, что каждый из них с равной вероятностью может выйти на любом из этажей, начиная со второго. Найти вероятность того, что:
а) все пятеро выйдут на пятом этаже;
б) все пятеро выйдут одновременно (на одном и том же этаже);
в) все пятеро выйдут на разных этажах.
```

````python
import numpy as np


# Параметры задачи
floors = list(range(2, 10))  # этажи 2-9
n_people = 5
n_trials = 1000000

a_success = 0  # для пункта (а)
b_success = 0  # для пункта (б)
v_success = 0  # для пункта (в)

np.random.seed(42)

for _ in range(n_trials):
    choices = np.random.choice(floors, n_people)

    # (a): все пятеро вышли на пятом этаже:
    if np.all(choices == 5):
        a_success += 1

    # (b): все вышли одновременно (на одном этаже)
    if len(set(choices)) == 1:
        b_success += 1

    # (в): все вышли на разных этажах
    if len(set(choices)) == n_people:
        v_success += 1

print(f"Вероятность что все выйдут на пятом этаже: {a_success / n_trials:.7f}")
print(f"Вероятность что все выйдут одновременно: {b_success / n_trials:.7f}")
print(f"Вероятность что все выйдут на разных этажах: {v_success / n_trials:.7f}")

````

Аналитические решения:
```
а)0.0000305
б)0.000244
в)0.205
```



Задача 3
```
Случайная точка A наудачу выбирается в прямоугольнике со
сторонами 1 и 2. Найти вероятности событий: расстояние от A

а) до ближайшей стороны прямоугольника не превосходит x;

б) до каждой стороны прямоугольника не превосходит x;

в) до каждой диагонали прямоугольника не превосходит x.

```



````python
import numpy as np

# Параметры задачи
a = 1   # ширина прямоугольника
b = 2   # высота прямоугольника
x = 0.15 # интересующий отрезок: 0 < x < 0.5
N = 1000000  # число испытаний для Монте-Карло

np.random.seed(42)

# Генерация случайных точек
points = np.empty((N, 2))
points[:,0] = np.random.uniform(0, a, N)
points[:,1] = np.random.uniform(0, b, N)

# (а) Расстояние до ближайшей стороны <= x
dists_nearest = np.min([points[:,0], a-points[:,0], points[:,1], b-points[:,1]], axis=0)
pa_mc = np.mean(dists_nearest <= x)
# Аналитика
if x >= min(a, b)/2:
    pa_an = 1.0
else:
    pa_an = (2*a + 2*b)*x - 4*x**2
    pa_an /= (a*b)

# (б) Расстояние до каждой стороны <= x
mask_inside = (
    (points[:,0] >= x) & (points[:,0] <= a - x)
    & (points[:,1] >= x) & (points[:,1] <= b - x)
)
pb_mc = np.mean(mask_inside)
# Аналитика
if x > min(a, b)/2:
    pb_an = 0.0
else:
    pb_an = ((a - 2*x) * (b - 2*x)) / (a*b)

# (в) До каждой диагонали <= x
# Диагональ 1: y = (b/a)*x
# Диагональ 2: y = -(b/a)*x + b
sqrt_k = np.sqrt((b/a)**2 + 1)
d1 = np.abs((b/a)*points[:,0] - points[:,1]) / sqrt_k
d2 = np.abs(-(b/a)*points[:,0] - points[:,1] + b) / sqrt_k
mask_diag = (d1 <= x) & (d2 <= x)
pc_mc = np.mean(mask_diag)
# Аналитика: нет простого явного выражения

# Вывод результата
print(f"""\
Расстояние до ближайшей стороны ≤ x:
  Монте-Карло: {pa_mc:.5f}

Расстояние до каждой стороны ≤ x:
  Монте-Карло: {pb_mc:.5f}

Расстояние до каждой диагонали ≤ x:
  Монте-Карло: {pc_mc:.5f}
""")
````
